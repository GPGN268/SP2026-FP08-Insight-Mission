# Frequency vs Distance 

```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def load_marsquakes_df(filename):
    # Column names based on your file structure: 
    # ID, Timestamp, Distance, Quality, Class
    cols = ['Event_ID', 'Timestamp', 'Distance', 'Quality', 'Class']
    
    # read_csv with sep='\s+' handles multiple spaces
    # on_bad_lines='skip' handles the markers in your text files
    df = pd.read_csv(filename, sep=',', names=cols, on_bad_lines='skip', skiprows= 1)

    df['Timestamp'] = pd.to_datetime(df['Timestamp'], format= 'ISO8601')
    
    return df

df_a = load_marsquakes_df('data/marsquakes_data_frame.csv')

df_a.head(10)
```

```python
df_a = df_a.sort_values('Timestamp')

df_a_filtered = df_a[df_a['Quality'] == 'A']
df_b_filtered = df_a[df_a['Quality'] == 'B']

quality= df_a.Quality
distance= df_a.Distance

fig, ax1= plt.subplots(figsize=(15,6))

color = 'deeppink'
line1= ax1.plot(df_a_filtered['Timestamp'], df_a_filtered['Distance'], color=color, linewidth=2, label='Quality A')
ax1.set_ylabel('Distance')
ax1.set_xlabel('Date of Event')
#ax1.set_ylim()
#ax1.invert_yaxis()
ax1.tick_params(axis='x', labelcolor=color)
ax1.grid(ls='dotted', which='both') #dotted grid lines

ax2 = ax1.twiny()
color = 'royalblue'
line2= ax2.plot(df_b_filtered['Timestamp'], df_b_filtered['Distance'], color=color, linewidth=2, label='Quality B')
#ax2.set_xlabel()
ax2.tick_params(axis='x', labelcolor=color)

lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Distance from Lander of Quality A and B Events')

plt.show()
```

**Preparing the Data:**

1) **Inital Survey:** A plot was made made of only the events in potential volatile-rich zones. This looked very similar to the plot A above, but inlcuded 2-3 oulier events that occurred beyond 1900 km from the lander.
2) **Data Truncation:** Both the volatile-rich dataset and the dataset including all A-B events were filtered to only include events between 1300 and 1900 km. Targeting specific data provided a closer look at the densest area of readings, allowing for better conclusions and analysis.
3) **Establishing a control:** A plot was made using all A-B events -- **this is the control group.** 

Using a control group is crucial to account for *seismic bias.* If a region is more seismically active in general, there will be a spike in both datasets. Without the control group, a spike in the volatile-rich plot could lead to the false conclusion that volatiles are uniquely concentrated there, when in reality we are simply seeing a higher volume of total seismic activity.

4) **Comparative Visualization:** Both datasets were plotted on top of each other to show devitations between the general seismic data and the volatile-rich subset.

**Analysis:**

The pink area represents a 95% confidence interval for where volatile-rich zones are predicted to occur. The purple area represents a 95% confidence interval for the total seismic baseline including all quality A-B events. The area under each curve is 1, which normalizes each curve to represent relative probability density rather than raw counts.

**We are 95% confident that the true distribution of volatile-rich zones lies within the shaded pink region, assuming the data represents a random sampling of marsquakes.**

Plotting both datasets on top of each other allows us to easily identify anomailes where the volatile-rich data diverges from the baseline readings.

Any place where both confidence intervals are not overlapping is considered statistically significant. The most obvious instance of this happening peaks around 1470 km, where the pink confidence interval far exceeds the purple baseline, showing that you are more likely to find a volatile-rich zone in that area than a normal, dry-rock zone. While there is a spike in overall marsquakes at this distance, the disproportionate concentration of volatile-rich signals suggest an area containing volatiles.



```python
def load_final_marsquakes_df(filename):
    
    cols = ['Event_ID', 'Distance_Degrees', 'Distance_KM', 't0', 'tp', 'ts', 'Vp_Vs_Ratio']
    
    
    df = pd.read_csv(filename, sep=',', names=cols, on_bad_lines='skip', skiprows= 1)

    return df


df_a = load_final_marsquakes_df('final_marsquakes_data_frame.csv')
volatile_zone = df_a[(df_a['Vp_Vs_Ratio'] >= 1.80) & (df_a['Vp_Vs_Ratio'] <= 2.10)].copy()

# depth is ~12% of the surface distance
volatile_zone['Est_Depth_KM'] = volatile_zone['Distance_KM'] * 0.12

volatile_zone.head(10)
```

```python
plt.figure(figsize=(10, 6))

n, bins, patches = plt.hist(volatile_zone['Distance_KM'], 
                            bins=30, 
                            color='skyblue', 
                            edgecolor='black', 
                            alpha=0.7)

plt.title('Frequency of Volatile-Rich Events by Distance', fontsize=14, pad=15)
plt.xlabel('Distance from Lander (KM)', fontsize=12)
plt.ylabel('Number of Events (Frequency)', fontsize=12)

plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

dist = volatile_zone['Distance_KM'].values

# 2. Setup Figure
fig, ax1 = plt.subplots(figsize=(10, 6))

n, bins, patches = ax1.hist(dist, bins=25, density=True, color='skyblue', 
                            edgecolor='black', alpha=0.5, label='Event Frequency')

kde = gaussian_kde(dist)
dist_range = np.linspace(min(dist), max(dist), 500) # Smooth X-axis
ax1.plot(dist_range, kde(dist_range), color='deeppink', linewidth=3, label='Trend Density')

ax1.set_title('Distribution Trend of Volatile-Rich Zones', fontsize=14)
ax1.set_xlabel('Distance from Lander (KM)')
ax1.set_ylabel('Probability Density')
ax1.legend()
ax1.grid(axis='y', ls='dotted', alpha=0.6)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

dist = volatile_zone['Distance_KM'].values
n = len(dist)
color = 'deeppink'
fig, ax1 = plt.subplots(figsize=(10, 6))

n_hist, bins, patches = ax1.hist(dist, bins=25, density=True, color='skyblue', 
                                 edgecolor='black', alpha=0.3, label='Event Frequency')

kde = gaussian_kde(dist)
dist_range = np.linspace(min(dist), max(dist), 500)
trend_density = kde(dist_range)

z_score = 1.96
bandwidth = kde.factor * np.std(dist) # Scipy's internal smoothing factor
std_err = np.sqrt(trend_density / (n * bandwidth))
ci = z_score * std_err

ax1.plot(dist_range, trend_density, color=color, linewidth=3, label='Smooth Trend')

ax1.fill_between(dist_range, 
                 np.maximum(0, trend_density - ci), 
                 trend_density + ci, 
                 color=color, alpha=0.2, label='95% Confidence Interval')

ax1.set_title('Smooth Distribution Trend of Volatile-Rich Zones', fontsize=14)
ax1.set_xlabel('Distance from Lander (KM)')
ax1.set_ylabel('Probability Density')
ax1.grid(ls='dotted', alpha=0.6)
ax1.legend()

plt.show()
```

```python
depth = volatile_zone['Est_Depth_KM'].values
n = len(depth)
color = 'deeppink'

fig, ax1 = plt.subplots(figsize=(10, 6))

n_hist, bins, patches = ax1.hist(depth, bins=25, density=True, color='skyblue', 
                                 edgecolor='black', alpha=0.3, label='Event Frequency')

kde = gaussian_kde(depth)
depth_range = np.linspace(min(depth), max(depth), 500)
trend_density = kde(depth_range)

z_score = 1.96
bandwidth = kde.factor * np.std(depth) # Scipy's internal smoothing factor
std_err = np.sqrt(trend_density / (n * bandwidth))
ci = z_score * std_err

ax1.plot(depth_range, trend_density, color=color, linewidth=3, label='Probability Density Estimate')

ax1.fill_between(depth_range, 
                 np.maximum(0, trend_density - ci), 
                 trend_density + ci, 
                 color=color, alpha=0.2, label='95% Confidence Interval')

ax1.set_title('Predicting Volatile-Rich Zones Based on Depth', fontsize=14)
ax1.set_xlabel('Depth (KM)')
ax1.set_ylabel('Probability Density')
ax1.grid(ls='dotted', alpha=0.6)
ax1.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

dist = volatile_zone['Distance_KM'].values

fig, ax1 = plt.subplots(figsize=(10, 6))

n_counts, bins, patches = ax1.hist(dist, bins=25, density=True, color='skyblue', 
                                   edgecolor='black', alpha=0.5, label='Event Frequency')

bin_centers = (bins[:-1] + bins[1:]) / 2

window = 3
trendline = np.convolve(n_counts, np.ones(window)/window, mode='same')

z_score = 1.96
std_err = np.std(n_counts) / np.sqrt(len(dist))
ci = z_score * std_err

color = 'deeppink'
ax1.plot(bin_centers, trendline, color=color, linewidth=3, label='Trend Density')

ax1.fill_between(bin_centers, 
                 np.maximum(0, trendline - ci),
                 trendline + ci, 
                 color=color, alpha=0.2, label='95% Confidence Interval')

ax1.set_title('Distribution Trend of Volatile-Rich Zones', fontsize=14)
ax1.set_xlabel('Distance from Lander (KM)')
ax1.set_ylabel('Probability Density')
ax1.legend()
ax1.grid(axis='y', ls='dotted', alpha=0.6)

plt.show()
```
